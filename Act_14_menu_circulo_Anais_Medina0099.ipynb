#### Redaccion y contexto del problema
Se requiere desarrollar un programa interactivo en Python que permita calcular propiedades geométricas básicas a partir del radio ingresado por el usuario. El programa debe mostrar un menú con cuatro opciones: calcular el área de un círculo, calcular su perímetro, calcular el volumen de una esfera y salir del sistema. El programa debe repetir su ejecución mediante un bucle hasta que el usuario decida terminar, validando las opciones ingresadas y utilizando funciones separadas para cada cálculo.

<img src="https://raw.githubusercontent.com/AnaisMedinaZubia/Mis-imagenes-de-EPPL-y-trabajos/refs/heads/main/circulo.PNG" width="300" height="500" alt="Texto descriptivo">

```python
import math

def areacirculo():
    """Calcula el área de un círculo a partir del radio ingresado."""
    radio = float(input("\nIngresa el radio del círculo: "))
    area = math.pi * (radio ** 2)
    print(f"✅ El área del círculo es: {area:.2f}")

def perimetrocirculo():
    """Calcula el perímetro (circunferencia) de un círculo."""
    radio = float(input("\nIngresa el radio del círculo: "))
    perimetro = 2 * math.pi * radio
    print(f"✅ El perímetro del círculo es: {perimetro:.2f}")

def volumencirculo():
    """Calcula el volumen de una esfera.
    (Se mantiene el nombre solicitado en la consigna)"""
    radio = float(input("\nIngresa el radio de la esfera: "))
    volumen = (4/3) * math.pi * (radio ** 3)
    print(f"✅ El volumen de la esfera es: {volumen:.2f}")

def menu():
    """Muestra el menú principal y controla el flujo del programa."""
    while True:
        print("\n" + "="*30)
        print(" MENÚ DE CÁLCULOS GEOMÉTRICOS 🔵")
        print("="*30)
        print("1. Calcular área del círculo")
        print("2. Calcular perímetro del círculo")
        print("3. Calcular volumen de la esfera")
        print("4. Salir")
        print("="*30)

        opcion = input("Selecciona una opción (1-4): ")

        # Estructura de control if-elif-else
        if opcion == "1":
            areacirculo()          # Llamada a función
        elif opcion == "2":
            perimetrocirculo()     # Llamada a función
        elif opcion == "3":
            volumencirculo()       # Llamada a función
        elif opcion == "4":
            print("\n👋 ¡Gracias por usar el programa! Hasta luego.")
            break                  # Sale del bucle while
        else:
            print("⚠️ Opción no válida. Por favor, ingresa un número del 1 al 4.")

# Punto de entrada del programa
if __name__ == "__main__":
    menu()
```
